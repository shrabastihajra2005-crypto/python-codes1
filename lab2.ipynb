```python
from collections import deque

def water_jug_bfs():
    capacity_a = 4
    capacity_b = 3
    start = (0, 0)

    queue = deque([(start, [])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        # Goal: exactly 2 litres in either jug
        if a == 2 or b == 2:
            return path + [(a, b)]

        next_states = []

        # 1. Fill Jug A
        next_states.append((capacity_a, b))

        # 2. Fill Jug B
        next_states.append((a, capacity_b))

        # 3. Empty Jug A
        next_states.append((0, b))

        # 4. Empty Jug B
        next_states.append((a, 0))

        # 5. Pour A -> B
        pour = min(a, capacity_b - b)
        next_states.append((a - pour, b + pour))

        # 6. Pour B -> A
        pour = min(b, capacity_a - a)
        next_states.append((a + pour, b - pour))

        for state in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b)]))

    return None


solution = water_jug_bfs()

print("Minimum sequence of states:")
for i, state in enumerate(solution):
    print(f"Step {i}: {state}")
```
